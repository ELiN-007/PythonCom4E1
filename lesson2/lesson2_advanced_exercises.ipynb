## Exercise 1: Numeric analysis

- Create one `int`, one `float` (scientific notation), and one `complex` number.

## Exercise 2: Safe casting challenge

- Given `values = ["25", "3.14", "-8", "hello"]`, decide how each text should be converted.
- Use string methods (such as `isdigit()`, `replace()`, `lstrip()`) and `if/elif/else` to check each value.
- Convert to `int` if it is a whole number, to `float` if it is a decimal number, otherwise keep it as a string.
- Print each converted value together with its type.

```python
values = ["25", "3.14", "-8", "hello"]

for val in values:
    # Check if it's an integer (handling potential negative sign)
    is_int_str = val.lstrip('-').isdigit()
    
    # Check if it's a decimal/float number (single dot, digits on both sides, optional negative sign)
    cleaned = val.lstrip('-')
    parts = cleaned.split('.')
    is_float_str = len(parts) == 2 and parts[0].isdigit() and parts[1].isdigit()
    
    # Conditional conversion
    if is_int_str:
        converted = int(val)
    elif is_float_str:
        converted = float(val)
    else:
        converted = val
        
    print(f"Value: {val:<8} | Converted: {str(converted):<8} | Type: {type(converted)}")
```

## Exercise 3: String slicing matrix

- Use `text = "DataSciencePython"`.
- Print slices: first 4 chars, last 6 chars, every 2nd char, reversed text.
- Extract and print `Science` from the string using slicing only.

```python
# Write your solution here
text = "DataSciencePython"

# 1. First 4 characters
first_4 = text[:4]

# 2. Last 6 characters
last_6 = text[-6:]

# 3. Every 2nd character
every_2nd = text[::2]

# 4. Reversed text
reversed_text = text[::-1]

# 5. Extract "Science" using slicing only
science = text[4:11]

# Print results
print(f"First 4 chars:    {first_4}")
print(f"Last 6 chars:     {last_6}")
print(f"Every 2nd char:   {every_2nd}")
print(f"Reversed text:    {reversed_text}")
print(f"Extracted word:   {science}")
```

## Exercise 4: String transformation pipeline

- Start with `raw = "   python,java,c++ , go  "`.
- Strip the outer spaces and split the text by comma.
- Print the split list.
- Loop through the list, strip each language, uppercase it, and print it.

```python
# Write your solution here
raw = "    python,java,c++ , go   "

# 1. Strip the outer spaces and split the text by comma
cleaned_raw = raw.strip()
split_list = cleaned_raw.split(',')

print("Split list:", split_list)

# 2. Loop through the list, strip each language, uppercase it, and print
print("\nProcessed languages:")
for lang in split_list:
    formatted_lang = lang.strip().upper()
    print(formatted_lang)
```

## Exercise 5: Formatting and escaping

- Create variables `product`, `price`, and `discount`.
- Print original and discounted price using f-string with 2 decimals.
- Print one line containing escaped quotes and another with newline+tab formatting.

```python
# Write your solution here
# Variables
product = "Laptop"
price = 1250.5
discount = 15.0  # percentage

# Calculate discounted price
discounted_price = price * (1 - discount / 100)

# Print original and discounted price using f-strings with 2 decimals
print(f"Product: {product}")
print(f"Original Price:   ${price:.2f}")
print(f"Discounted Price: ${discounted_price:.2f}")

# Print line with escaped quotes
print("\nEscaped quotes example:")
print("She exclaimed, \"Learning Python is fun!\"")

# Print line with newline and tab formatting
print("\nNewline and tab formatting example:")
print("Inventory:\n\tCategory: Hardware\n\tItem: Laptop")
```
