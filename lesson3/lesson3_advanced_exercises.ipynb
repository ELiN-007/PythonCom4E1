## Exercise 1: Truth table mini-lab

- Create booleans `p = True`, `q = False`.
- Print results for: `p and q`, `p or q`, `not p`, `not q`.
- Add one compound expression with parentheses and print its result.

```python
# Write your solution here
p = True
q = False
print(f"p and q is {p and q}")
print(f"p and q is {p or q}")
print(f"not p is {not p}")
print(f"not q is {not q}")

compound_result = (p or q) and not (p and q)
print(f"(p or q) and not (p and q): {compound_result}")
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
# Write your solution here

list1 = [1, 2, 3]
list2 = [1, 2, 3]

print("Comparing separate lists")
print(f"list1 == list2 (Equality): {list1 == list2}")
print(f"list1 is list2 (Identity): {list1 is list2}")

print()


list3 = list1

print("--- Comparing assigned lists ---")
print(f"list1 == list3 (Equality): {list1 == list3}")
print(f"list1 is list3 (Identity): {list1 is list3}")
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
# Write your solution here
items = ["python", "java", "go", "rust"]

is_python_present = "python" in items
is_javascript_absent = "javascript" not in items
is_rust_present = "rust" in items
is_cpp_absent = "c++" not in items

print(f"python: {'found' if is_python_present else 'not found'}")
print(f"javascript: {'not found' if is_javascript_absent else 'found'}")
print(f"rust: {'found' if is_rust_present else 'not found'}")
print(f"c++: {'not found' if is_cpp_absent else 'found'}")
```
