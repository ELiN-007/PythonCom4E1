## Exercise 1: Nested list processing

- Create a 2D list (3 rows x 4 cols) of numbers.
- Print row sums and grand total.
- Print the maximum value from each row.

```python
matrix = [
    [3, 8, 1, 6],
    [5, 2, 9, 4],
    [7, 0, 3, 5]
]

grand_total = 0

for row in matrix:
    sum_row = 0
    for col in row:
        sum_row += col
    grand_total += sum_row
    print(sum_row)

print(f"grand total: {grand_total}")

for row in matrix:
    max_val = row[0]
    for col in row:
        if col > max_val:
            max_val = col
    print(max_val)
```

## Exercise 2: Comprehension with condition

- From numbers 1..30, build a list of numbers divisible by 3 but not by 2.
- Then build a second list of their squares using comprehension.
- Print both lists.

```python
numbers = []
for num in range(1, 31):
    if num % 3 == 0 and num % 2 != 0:
        numbers.append(num)

squares = []
for num in numbers:
    squares.append(num ** 2)

print("Numbers:", numbers)
print("Squares:", squares)
```

```python
# Write your solution here
list = [
    ["Alice", 85],
    ["Bob", 92],
    ["Charlie", 85],
    ["Diana", 95],
    ["Eve", 78]
]

def sort_key(student):
    return [-student[1], student[0]]

list.sort(key=sort_key)

print(list)
```
